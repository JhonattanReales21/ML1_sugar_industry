**Comments**

--> We have a **total number of 85 variables**. According to the context, we can note groups of variables, for example:
- Farm and soil information
- Crop characteristics
- Crop maturity
- Before harvest
- Irrigation characteristics
- Weather conditions 
- Fertilization and supplements
- And finally, we have several metrics of crop performance (**after harvest**), i.e., TCH, sucrose percentage, among others.

--> The **columns have different formats**, some were identified as datetime type, others as integer or decimal type and others as object/categorical type variables. We will modify some columns formats in next steps.

--> As for the **presence of null values**, we noted variables with more than 80% of null observations and others with less than 10%. We can observe that groups of variables present a very similar amount of nulls, for example, the variables of crop maturity present a percentage close to 47%, variables of applied fertilizers have a percentage of more than 80% and variables of meteorological conditions present a percentage of nulls of approximately 78.5%. This should be taken into account when making decisions on the inclusion or elimination of variables.

**Cleaning actions**
1. We must keep in mind that the variables to be studied and predicted are only two: Tons of cane per hectare (TCH) and sucrose percentage of cane harvested (pct_sac_cane).  
Therefore, **all the other variables of the harvest performance metrics group should be removed**, in order to **avoid problems such as data leakage** or similar.  
Also, the variables that could be measure only after harvests, should be removed.
2. We will drop he columns with %100 of nan values (sum_oscilacion_temp_ciclo, fert_nitrogen_)


#### cleaning

```python
var_to_remove = [
    "tonultcorte",
    "tchm",
    "ton_azucar",
    "rdto",
    "tah",
    "tahm",
    "sac_caña_precosecha",
    "pct_sac_muestreadora",
    "pct_atr",
    "katrhm",
    "pct_fibra_caña",
    "pct_ar_jugo",
    "pct_me_min",
    "pct_me_veg",
    "pct_me_tot",
    "brix",
    "pureza",
]

# drop columns that are metrics after harvest
df = df.drop(var_to_remove, axis=1)

# drop columns with only NAN values
df = df.drop(["sum_oscilacion_temp_ciclo", "fert_nitrogen_"], axis=1)

print(f'Number of columns: {df.shape[1]}')
```

### Identification - obs identifier

--> This exploration is **to find any possible duplicated observation**, and to find out what could be the identification of one valid observation.

```python
print(f"A farm (hacienda) could have more than one harvest period:\n")
print(
    df.groupby(by="hacienda")
    .agg({"período": lambda x: x.nunique()})
    .sort_values(by="período", ascending=False)
    .rename(columns={"período": "#periodos"})
)
```

```python
print(
    f"A farm (hacienda) could have one or two different zones (zona):\ni.e. Farm id #80134 zones:\n"
)
print(df[(df["hacienda"] == 80134)].zona.value_counts())
print(f"\nGeneral:\n")
print(
    df.groupby(by="hacienda")
    .agg({"zona": lambda x: x.nunique()})
    .sort_values(by="zona", ascending=False)
    .rename(columns={"zona": "#zonas"}).iloc[5:15,:]
)
```

```python
print(
    f"A combination of farm-zone could have one or two different 'tenencia' values:\n"
)
df.groupby(by=["hacienda", "zona"]).agg(
    {
        "tenencia": [
            ("#", lambda x: x.nunique()),
            ("names", lambda x: x.unique().tolist()),
        ]
    }
).sort_values(by=[("tenencia", "#"), "hacienda"], ascending=[False, False]).iloc[10:20:,:]
```

```python
print(
    f"Moreover, a combination of farm-zone-tenure could have more than one plot (suerte):\n"
)

df.groupby(by=["hacienda", "zona","tenencia"]).agg(
    {
        "suerte": [
            ("#", lambda x: x.nunique()),
            ("names", lambda x: x.unique().tolist()),
        ]
    }
)
```

```python
print(
    f"So, taking in consideration the analysis, we could define our observations ID like the combination of period-farm-zone-tenure-plot:\n"
)
df.groupby(by=["período","hacienda", "zona","tenencia","suerte"]).count()["tch"].sort_values()
```

```python
print(
    f"And also, we could determine the harvests carried out in the same plot or same soil, with the combination of the variables farm-zone-tenure-plot \n"
)
df.groupby(by=["hacienda", "zona","tenencia","suerte"]).count()["tch"].sort_values()
```

**Cleaning actions**
1. We will drop the "nombre" column, because this variable do not give valuable information and have some typos.
2. We will modify the "zona" levels, lowering the letters.
3. All variables belonging to the observation identifier, will be cast to object format.
4. We will modify the "suelo" levels, lowering the letters and replacing the spaces with underscore.
5. We will sort the dataframe according to the identifier of obs.


```python
# drop column "nombre"
df = df.drop("nombre",axis=1)

# lower the levels of variable "zona"
df.zona = df.zona.str.lower()

# Cast hacienda & tenencia to object format
df[["hacienda","tenencia"]] = df[["hacienda","tenencia"]].astype(object)

# modify the levels of "suelo"
df.suelo = df.suelo.str.strip().str.lower().str.replace(" ","_")

# sort the dataframe
df = df.sort_values(by=["período","hacienda","zona","suerte","f_siembra"], ascending=[True,True,True,True,True]).reset_index(drop=True)
```
