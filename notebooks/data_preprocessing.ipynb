**Comments**
 
--> we can notice a number of harvests with a **very old sowing date**.  
--> We can notice some crops with **very extreme crop ages (edad_ult_cos)**, these should be analyzed in detail in the EDA.  
--> The columns cod_estado_# & cod_estado refer to the same thing. The same case for the columns cod__t_cultivo & cultivo.  
--> **When the crop is destined for seed, there are no sucrose percentage metrics.** There are also changes in different variables and some columns are NaN values.  
--> We consider that **the column “variedad” is very important for the analysis**, however, we must define how we are going to work with it, if we are going to categorize it or similar.

**Cleaning actions**
1. We will **drop the very old sowing dates with low harvest cuts** to avoid any potential bias in the analysis.  
2. We will **drop the column cod_estado and rename cod_estado_# to num_cosechas**.  
3. We will **drop the column cultivo and rename cod__t_cultivo to cultivo_organico**. After that, substract 1 from the values. 
4. We are going to **eliminate the crops that were intended for seed**. We will assume that we are only working with crops that are intended for production. Also, we will **drop the column destino_1=semilla** .  
5. We will **drop the column d_s_** because it is not relevant for the analysis and inference of tch.  
6. We will **rename the levels of column "variedad"** in order to make it more understandable and easier to work with.

```python
# Eliminate the old sowing dates with low harvest cuts
df = df[~((df.f_siembra < df.f_siembra.quantile(0.025)) & (df["cod_estado_#"] < 8))]

# drop cod_estado and rename cod_estado_#
df.drop("cod_estado", axis=1, inplace=True)
df.rename(columns={"cod_estado_#": "num_cosechas"}, inplace=True)

# drop cultivo and rename cod__t_cultivo, later, subtract 1 from the values
df.drop("cultivo", axis=1, inplace=True)
df.rename(columns={"cod__t_cultivo": "cult_organico"}, inplace=True)
df["cult_organico"] = df["cult_organico"] - 1

# Eliminate crops for seed and drop column destino_1=semilla
df = df[~(df["destino_1=semilla"] == 1)]
df.drop("destino_1=semilla", axis=1, inplace=True)

# Drop column d_s_
df.drop("d_s_", axis=1, inplace=True)

# rename the levels of "variedad"
df["variedad"] = df["variedad"].str.lower().str.replace("-", "_")
```

**Comments**
 
--> There are a large number of observations with no data on when the last water irrigation took place. In these cases, the other variables related to irrigation will take the value of zero for that crop.  
--> Observations without rainfall data take the value of zero for all columns.  

**Cleaning actions**
1. The name of the column “luvias_9__fc” will be changed to “lluvia_9_a_cos”.  
  
*With a view to perform imputation steps in the future:*  

2. Observations with a null value in the date of the last irrigation (ult_riego), a null value will be placed in "num_riegos","ddult_riego" and "m3_riego".  
3. Observations with zero values in "lluvias_ciclo", will be changed by a null value.     
4. After the step 3, the observations with null value in "lluvias_ciclo", a null value will be placed in the other rainfall variables.  

```python
# change coumn name for luvias_9__fc
df.rename(columns={"luvias_9__fc": "lluvias_9_a_cos"}, inplace=True)

# place null values for irrigation variables
df.num_riegos = np.where(df.ult_riego.isna(), np.nan, df.num_riegos)
df.ddult_riego = np.where(df.ult_riego.isna(), np.nan, df.ddult_riego)
df.m3_riego = np.where(df.ult_riego.isna(), np.nan, df.m3_riego)

# place null values in column "lluvias_ciclo"
df.lluvias_ciclo = np.where(df.lluvias_ciclo == 0, np.nan, df.lluvias_ciclo)

# place null values for other rainfall columns
df.lluvias_2_meses_ant_ = np.where(
    df.lluvias_ciclo.isna(), np.nan, df.lluvias_2_meses_ant_
)
df.lluvias_0__3 = np.where(df.lluvias_ciclo.isna(), np.nan, df.lluvias_0__3)
df.lluvias_tres_a_seis = np.where(
    df.lluvias_ciclo.isna(), np.nan, df.lluvias_tres_a_seis
)
df.lluvias_seis_a_nueve = np.where(
    df.lluvias_ciclo.isna(), np.nan, df.lluvias_seis_a_nueve
)
df.lluvias_9_a_cos = np.where(df.lluvias_ciclo.isna(), np.nan, df.lluvias_9_a_cos)
```
